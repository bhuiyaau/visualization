```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1=np.random.randint(0,100,50)
y2=np.random.randint(0,100,50)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='Person1')
ax.plot(x,y2, label='Person2')
ax.legend(loc='lower right', frameon=True,shadow=True,fontsize=12,ncol=2)
#frameon= frame around the legend
#fontsize= change font size
#ncol= specify number of columns
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label='Person1')
ax.plot(x,y2, label='Person2')
ax.legend(loc='upper left', bbox_to_anchor=(1,1))
#bbox_to_anchor = moves outside of the graph
```

```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1=np.arange(50)
y1=np.random.randint(0,75,50)
x2=np.array(["luffy", "Zoro", "Nami","Usopp", "Sanji"])
y2=np.array([110,180,240,99,220])
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,nrows=1,figsize=(7,3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
ax1.scatter(x1,y1,marker='*',color='indigo')
ax2.bar(x2,y2,color='orange')
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,nrows=1,figsize=(7,3))
ax1.scatter(x1,y1)
ax1.annotate("This is important",xy=(10,60),xytext=(20,60),arrowprops=dict(facecolor='black'))
ax2.bar(x2,y2)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],['ax2', 'ax3']],figsize=(7.4, 5),layout="constrained")
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18)
```

```python
x=np.array(["Luffy", "Zoro", "Nami","Usopp", "Sanji"])
y1=np.array([110,180,240,99,220])
y2=np.array([170,100,90,120,50])
#define our figure and axes (just one this time)
fig,ax=plt.subplots(figsize=(7,3))

#now call both bar and plot elements on the same axes (ax)
ax.bar(x,y1, color="indigo")
ax.plot(x,y2, color="red")
```

```python
y2_sd=np.std(y2)
```

```python
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color='red')
```

```python
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color='red')
ax.errorbar(x,y2,yerr=y2_sd,fmt='o-',ecolor='indigo',elinewidth=4,capsize=6,capthick=4,errorevery=2)
```

```python
from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"
headers = {"User-Agent": "Mozilla/5.0","Accept": "image/*,*/*;q=0.8","Referer": "https://en.wikipedia.org/"}
response = requests.get(url, headers=headers)
image_file = BytesIO(response.content)
image = Image.open(image_file)
image
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax_image = fig.add_axes([0.1,0.11,0.15,0.35])
ax_image.imshow(image)
ax_image.axis('off')
```
